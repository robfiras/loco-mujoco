```python
import os
os.environ["MUJOCO_GL"] = "osmesa"  # for software rendering 

import time 
import jax 
import mujoco
import mujoco.viewer
from loco_mujoco import ImitationFactory
from loco_mujoco.environments.humanoids.skeletons import MjxSkeletonMuscle
import mediapy as media


spec = mujoco.MjSpec.from_file("/home/nadinebadie/loco-mujoco/loco_mujoco/models/skeleton/skeleton_muscle.xml")

getattr(spec.visual, "global").offwidth = 800
getattr(spec.visual, "global").offheight = 600

# # set visual offwidth and offheight to 800 and 600 respectively 
# spec.visual.global.offwidth = 800
# spec.visual.global.offheight = 600

model = spec.compile() #mujoco.MjModel.from_xml_path(xml_path)
data = mujoco.MjData(model)

```

```python
scene_option = mujoco.MjvOption()
# scene_option.flags[mujoco.mjtVisFlag.mjVIS_JOINT] = True
# scene_option.flags[mujoco.mjtVisFlag.mjVIS_TRANSPARENT] = True


# increase size of the rendered image
# scene_option.flags[mujoco.mjtVisFlag.SHADOW] = True


renderer = mujoco.Renderer(model) #, height= 480, width = 640)#, height=600, width=800)

duration = 0.2 #3.8  # (seconds)
framerate = 60  # (Hz)

frames = []
mujoco.mj_resetData(model, data)
while data.time < duration:
    mujoco.mj_step(model, data)
    if len(frames) < data.time * framerate:
        renderer.update_scene(data, scene_option=scene_option)
        pixels = renderer.render()
        frames.append(pixels)

# Simulate and display video.
media.show_video(frames, fps=framerate)
```
